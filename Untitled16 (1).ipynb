```python
from ultralytics import YOLO

# Load smallest model
model = YOLO("yolov8n.pt")

# Train (low epochs)
model.train(
    data="/content/pklot/data.yaml",
    epochs=3,        # 🔥 very low for speed
    imgsz=640,
    batch=8,
    name="parking_model"
)
```

```python
results = model.predict(
    source="/content/pklot/test/images",
    save=True,
    conf=0.25
)
```

```python
from IPython.display import Image, display
import glob

images = glob.glob("/content/runs/detect/predict/*.jpg")

for img in images[:5]:
    display(Image(filename=img))
```

```python
def count_parking(results):
    empty = 0
    occupied = 0

    for r in results:
        for box in r.boxes:
            cls = int(box.cls)

            if cls == 0:
                empty += 1
            else:
                occupied += 1

    print("Empty:", empty)
    print("Occupied:", occupied)

count_parking(results)
```
